```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

X_train = np.array([[25, 40000],
                    [30, 55000],
                    [45, 80000],
                    [50, 95000]])
X_test = np.array([[28, 50000],
                   [60, 110000]])

scaler = MinMaxScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
print(f"training data transformed:\n{X_train}")
print(f"test data transformed:\n{X_test}")
```
