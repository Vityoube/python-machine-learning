```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

iris = load_iris()
x = iris.data
y = iris.target

model = DecisionTreeClassifier()
model.fit(x, y)

new_flowers = [
    [5.1, 3.5, 1.4, 0.2],
    [6.7, 3.0, 5.2, 2.3],
    [5.9, 3.0, 4.2, 1.5]
]
predictions = model.predict(new_flowers)
print(predictions)
```
