```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

# Load the data
iris = load_iris()
X = iris.data    # flower measurements (features)
y = iris.target  # species: 0, 1, or 2 (target)

# Step 1: create a model and train it
model = DecisionTreeClassifier()
model.fit(X, y)

# Step 2: predict species for a new flower
new_flower = np.array([[5.0, 3.4, 1.5, 0.2]])
prediction = model.predict(new_flower)
print(prediction)  # [0] - the model thinks it is species 0 (setosa)

model = KNeighborsClassifier()
model.fit(X, y)
prediction = model.predict(new_flower)
print(prediction) # [0] - the model thinks it is species 0 (setosa)
```
