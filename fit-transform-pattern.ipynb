```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

X = np.array([[25, 30000],
              [45, 80000],
              [35, 55000],
              [50, 120000]])

scaler = MinMaxScaler()
scaler.fit(X)
X_scaled = scaler.transform(X)
print(X_scaled)

# [[0.   0.        ]
#  [0.8  0.55555556]
#  [0.4  0.27777778]
#  [1.   1.        ]]
```
